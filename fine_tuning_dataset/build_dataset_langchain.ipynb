```python
import json
from langchain.document_loaders import PyPDFLoader
from langchain.prompts import PromptTemplate
from langchain.llms.ollama import Ollama
```

```python
def load_pdf(path):
    loader = PyPDFLoader(path)
    pages = loader.load()
    return "\n".join([page.page_content for page in pages])

# Define prompt template
template = """
You are an expert in university-level exam question creation.
Generate a new question paper based on the study material, following the same structure and difficulty of the previous question paper.

### Previous Exam Paper:
{previous}

### Study Material:
{study}

📝 Generate the question paper below:
"""

prompt = PromptTemplate.from_template(template)

# Connect to Ollama's Mistral
llm = Ollama(model="mistral")

# Load PDFs
study_path = input("📘 Enter path to Study Material PDF: ").strip()
prev_path = input("📄 Enter path to Previous Question Paper PDF: ").strip()

study_text = load_pdf(study_path)
prev_text = load_pdf(prev_path)

# Format prompt
formatted_prompt = prompt.format(previous=prev_text, study=study_text)

# Get response from Mistral
print("⏳ Generating...")
response = llm.invoke(formatted_prompt).strip()

# Save to JSONL
entry = {
    "prompt": formatted_prompt,
    "response": response
}

with open("train1.jsonl", "a", encoding="utf-8") as f:
    f.write(json.dumps(entry, ensure_ascii=False) + "\n")

print("\n✅ Saved to train1.jsonl")
```
